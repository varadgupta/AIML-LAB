```python
light = False
fan = False
door = False
ac = False


def control_light(action):
    global light

    if action == "on":
        if light:
            print("Light is already ON")
        else:
            light = True
            print("Light turned ON")

    elif action == "off":
        if not light:
            print("Light is already OFF")
        else:
            light = False
            print("Light turned OFF")


def control_fan(action):
    global fan

    if action == "on":
        if fan:
            print("Fan is already ON")
        else:
            fan = True
            print("Fan turned ON")

    elif action == "off":
        if not fan:
            print("Fan is already OFF")
        else:
            fan = False
            print("Fan turned OFF")


def control_door(action):
    global door

    if action == "open":
        if door:
            print("Door is already OPEN")
        else:
            door = True
            print("Door OPENED")

    elif action == "close":
        if not door:
            print("Door is already CLOSED")
        else:
            door = False
            print("Door CLOSED")


def control_ac():
    global ac

    temp = float(input("Enter temperature: "))

    if temp > 30:
        if ac:
            print("AC is already ON")
        else:
            ac = True
            print("AC turned ON")

    elif temp < 30:
        if not ac:
            print("AC is already OFF")
        else:
            ac = False
            print("AC turned OFF")

    else:
        print("Temperature is 30°C. No change to AC.")


# Main program
while True:

    print("\n1. Turn on light")
    print("2. Turn off light")
    print("3. Turn on fan")
    print("4. Turn off fan")
    print("5. Open door")
    print("6. Close door")
    print("7. Turn AC on/off on temperature")
    print("8. Exit")

    choice = int(input("Enter your choice: "))

    match choice:
        case 1:
            control_light("on")

        case 2:
            control_light("off")

        case 3:
            control_fan("on")

        case 4:
            control_fan("off")

        case 5:
            control_door("open")

        case 6:
            control_door("close")

        case 7:
            control_ac()

        case 8:
            print("Program ended")
            break

        case _:
            print("Invalid choice")
```
