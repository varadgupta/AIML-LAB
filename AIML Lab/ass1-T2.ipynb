```python
while True:

    print("-IRRIGATION SYSTEM-")
    print("1. Soil Moisture")
    print("2. Water Pump")
    print("3. Rain Detection")
    print("4. Water Level")
    print("5. Irrigation Status")
    print("6. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        moisture = int(input("Enter soil moisture (%): "))

        if moisture < 30:
            print("Soil is dry")
        elif moisture > 70:
            print("Soil is wet")
        else:
            print("Soil moisture is normal")

    elif choice == 2:
        pump = input("Turn ON the water pump? (yes/no): ")

        if pump == "yes":
            print("Water Pump is ON")
        else:
            print("Water Pump is OFF")

    elif choice == 3:
        rain = input("Is it raining? (yes/no): ")

        if rain == "yes":
            print("Rain detected")
            print("Irrigation is stopped")
        else:
            print("No rain detected")
            print("Irrigation can continue")

    elif choice == 4:
        water = int(input("Enter water tank level (%): "))

        if water < 20:
            print("Water level is LOW")
        elif water > 80:
            print("Water tank is FULL")
        else:
            print("Water level is NORMAL")

    elif choice == 5:
        moisture = int(input("Enter soil moisture (%): "))

        if moisture < 30:
            print("Irrigation is REQUIRED")
        elif moisture > 70:
            print("Irrigation is NOT REQUIRED")
        else:
            print("Irrigation is OPTIONAL")

    elif choice == 6:
        print("Irrigation System stopped")
        break

    else:
        print("Invalid choice.")
```
